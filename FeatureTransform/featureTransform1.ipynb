Hamming Distance

```python
a = np.array([1,0,1,1,1,0,1])
b = np.array([0,0,0,1,0,1,1])
ham_d = np.sum(a != b)
print(f"Haming distance: {ham_d}")
```

```python
#this is real hamming distance
a = np.array([1,1])
b = np.array([1,1])
ham_d = np.sum(a != b)
print(f"Haming distance: {ham_d}")
```

```python
a = np.array([1,0,0,0,0,1])
b = np.array([0,0,1,1,1,1])
ham_d2 = distance.hamming(a, b)
ham_d2
```
