```python
titanic_data = sns.load_dataset("titanic")
```

```python
titanic_data.drop('deck', axis=1, inplace=True)
```

```python
titanic_data.head(2)
```

```python
titanic_data['age'] = titanic_data['age'].fillna(titanic_data['age'].mean())
```

```python
titanic_data['embarked'] = titanic_data['embarked'].bfill()
```

```python
titanic_data['embark_town'] = titanic_data['embark_town'].bfill()
```

```python
titanic_data['age'] = minmax.fit_transform(titanic_data[['age']])
```
