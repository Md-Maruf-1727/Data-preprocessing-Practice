```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import seaborn as sns
```

```python
titanic_data = sns.load_dataset("titanic")
```

```python
titanic_data.head()
```

```python
titanic_data.shape
```

```python
titanic_data.isna().sum()
```

```python
titanic_data.drop('deck', axis=1, inplace=True)
```

```python
titanic_data.head(2)
```

```python
titanic_data['age'] = titanic_data['age'].fillna(titanic_data['age'].mean())
```

```python
titanic_data['embarked'] = titanic_data['embarked'].bfill()
```

```python
titanic_data['embark_town'] = titanic_data['embark_town'].bfill()
```

```python
titanic_data.isna().sum()
```

```python
titanic_data.head()
```

```python
minmax = MinMaxScaler()
```

```python
titanic_data['age'] = minmax.fit_transform(titanic_data[['age']])
```

```python
titanic_data['fare'] = minmax.fit_transform(titanic_data[['fare']])
```
