```python
one_hot_encoding = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
X_categorical = titan_dataset1[['sex']]
```

```python
encoded_data = one_hot_encoding.fit_transform(X_categorical)
```

```python
encoded_data
```

```python
encoded_titanic_dataset1 = pd.DataFrame(encoded_data, columns=one_hot_encoding.get_feature_names_out(['sex'])).astype('int64')
```

```python
encoded_titanic_dataset1 = pd.concat([titan_dataset1.drop('sex', axis=1), encoded_titanic_dataset1], axis=1)
```

```python
encoded_titanic_dataset1.head(1)
```

```python
Y_categorical = encoded_titanic_dataset1[['embarked']]
Y_categorical_fit = one_hot_encoding.fit_transform(Y_categorical)
encoded_titanic_dataset2 = pd.DataFrame(Y_categorical_fit, columns=one_hot_encoding.get_feature_names_out(['embarked'])).astype('int64')
encoded_titanic_dataset2 = pd.concat([encoded_titanic_dataset1.drop('embarked', axis=1), encoded_titanic_dataset2], axis=1)
encoded_titanic_dataset2.head(1)
```

```python
X_categorical_fit = one_hot_encoding.fit_transform(encoded_titanic_dataset2[['class']])
encoded_titanic_dataset3 = pd.DataFrame(X_categorical_fit, columns=one_hot_encoding.get_feature_names_out(['class'])).astype('int64')
encoded_titanic_dataset3 = pd.concat([encoded_titanic_dataset2.drop('class', axis=1), encoded_titanic_dataset3], axis=1)
encoded_titanic_dataset3.head()
```

```python
X_categorical_fit = one_hot_encoding.fit_transform(encoded_titanic_dataset2[['who']])
encoded_titanic_dataset3 = pd.DataFrame(X_categorical_fit, columns=one_hot_encoding.get_feature_names_out(['who'])).astype('int64')
encoded_titanic_dataset3 = pd.concat([encoded_titanic_dataset2.drop('who', axis=1), encoded_titanic_dataset3], axis=1)
encoded_titanic_dataset3.head()
```

```python
for col in encoded_titanic_dataset3.loc[:, ['adult_male', 'alone']]:
    encoded_titanic_dataset3[col] = encoded_titanic_dataset3[col].astype('int64')
```

```python
encoded_titanic_dataset3.head()
```

```python
fit_x_cat = one_hot_encoding.fit_transform(encoded_titanic_dataset3[['class']])
encoded_titanic_dataset4 = pd.DataFrame(fit_x_cat, columns=one_hot_encoding.get_feature_names_out(['class'])).astype('int64')
encoded_titanic_dataset4 = pd.concat([encoded_titanic_dataset3.drop('class', axis=1), encoded_titanic_dataset4], axis=1)
encoded_titanic_dataset4.head()
```

```python
fit_x_cat = one_hot_encoding.fit_transform(encoded_titanic_dataset3[['embark_town']])
encoded_titanic_dataset5 = pd.DataFrame(fit_x_cat, columns=one_hot_encoding.get_feature_names_out(['embark_town'])).astype('int64')
encoded_titanic_dataset5 = pd.concat([encoded_titanic_dataset4.drop('embark_town', axis=1), encoded_titanic_dataset5], axis=1)
encoded_titanic_dataset5.head()
```

```python
fit_x_cat = one_hot_encoding.fit_transform(encoded_titanic_dataset3[['alive']])
encoded_titanic_dataset6 = pd.DataFrame(fit_x_cat, columns=one_hot_encoding.get_feature_names_out(['alive'])).astype('int64')
encoded_titanic_dataset6 = pd.concat([encoded_titanic_dataset5.drop('alive', axis=1), encoded_titanic_dataset6], axis=1)
encoded_titanic_dataset6.head()
```
