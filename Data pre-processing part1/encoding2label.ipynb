```python
penguin_data_set['bill_length_mm'] = penguin_data_set['bill_length_mm'].fillna(penguin_data_set['bill_length_mm'].mean())
```

```python
penguin_data_set['bill_depth_mm'] = penguin_data_set['bill_depth_mm'].fillna(penguin_data_set['bill_depth_mm'].mean())
```

```python
penguin_data_set['flipper_length_mm'] = penguin_data_set['flipper_length_mm'].fillna(penguin_data_set['flipper_length_mm'].mean())
```

```python
penguin_data_set['body_mass_g'] = penguin_data_set['body_mass_g'].fillna(penguin_data_set['body_mass_g'].mean())
```

```python
penguin_data_set['sex'] = penguin_data_set['sex'].ffill()
```

```python
penguin_data_set.isna().sum()
```

```python
penguin_data_set.columns
```

```python
for col in penguin_data_set.loc[:,['species', 'island','sex']]:
    val = penguin_data_set[col].value_counts()
    print(val)
```

```python
for col in penguin_data_set.loc[:,['species','island','sex']]:
    uni = penguin_data_set[col].nunique()
    print(uni)
```

```python
for col in penguin_data_set.columns:
    if is_numeric_dtype(penguin_data_set[col]):
        continue
    else:
        uni = penguin_data_set[col].nunique()
        val = penguin_data_set[col].unique().tolist()
        if uni == 2:
            penguin_data_set[col] = penguin_data_set[col].replace(val,[0,1])
        elif uni == 3:
            penguin_data_set[col] = penguin_data_set[col].replace(val,[0,1,2])
```
