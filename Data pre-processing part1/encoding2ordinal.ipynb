```python
import pandas as pd
import numpy as np
import seaborn as sns
import requests
from io import StringIO
from sklearn.preprocessing import OrdinalEncoder
from pandas.api.types import is_numeric_dtype
```

```python
tip_data_set = sns.load_dataset('tips')
```

```python
tip_data_set.head()
```

```python
tip_data_set.isna().sum()
```

```python
ordinal_en = OrdinalEncoder()
```

```python
tip_data_set['sex'] = ordinal_en.fit_transform(tip_data_set[['sex']]).astype('int64')
```

```python
tip_data_set['sex'].value_counts()
```

```python
tip_data_set.head()
```

```python
smok_fit = ordinal_en.fit(tip_data_set[['smoker']])
```

```python
smok_fit
```

```python
tip_data_set['smoker'] = smok_fit.transform(tip_data_set[['smoker']]).astype('int64')
```

```python
peng_data_set = sns.load_dataset('penguins')
```

```python
peng_data_set.head()
```

```python
peng_data_set.isna().sum()
```

```python
peng_data_set['sex'] = peng_data_set['sex'].ffill()
```

```python
peng_data_set['bill_length_mm'] = peng_data_set['bill_length_mm'].fillna(peng_data_set['bill_length_mm'].mean())
```

```python
peng_data_set['bill_depth_mm'] = peng_data_set['bill_depth_mm'].fillna(peng_data_set['bill_depth_mm'].mean())
```

```python
peng_data_set['flipper_length_mm'] = peng_data_set['flipper_length_mm'].fillna(peng_data_set['flipper_length_mm'].mean())
```

```python
peng_data_set['body_mass_g'] = peng_data_set['body_mass_g'].fillna(peng_data_set['body_mass_g'].mean())
```

```python
peng_data_set.isna().sum()
```

```python
for col in peng_data_set.loc[:, ['species','island','sex']]:
    peng_data_set[col] = ordinal_en.fit_transform(peng_data_set[[col]]).astype('int64')
```
